```python
df['Review Rating']=df['Review Rating'].fillna(df['Review Rating'].median())
```

```python
df.isnull().sum()
```

```python
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(' ','_')
```

```python
df.columns
```

```python
df.rename(columns={'purchase_amount_(usd)': 'purchase_amount'}, inplace=True)
```

```python
#create a col age_group
labels=['Younger Adult','Adult','Middle_aged','Senior']
df['age_group']=pd.qcut(df['age'],q=4,labels=labels)
```

```python
df[['age','age_group']].head(10)
```

```python
#create col purchase_frequency_days
frequency_mapping={'Fortnightly':14,'Weekly':7,'Monthly':30,'Quarterly':90,'Bi-weekly':14,'Annually':365,'Every 3 Mon':90}
df['purchase_frequency_days']=df['frequency_of_purchases'].map(frequency_mapping)
```
